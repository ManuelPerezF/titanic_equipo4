```python
# Filtrar menores de edad
menores = df[df["Age"] < 18]

# Matriz de correlación entre supervivencia y número de hermanos
correlacion = menores[["Survived", "SibSp"]].corr()

print(correlacion)
```

```python
print(menores[["Survived", "SibSp"]].corr().loc["Survived", "SibSp"])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Filtrar menores de edad
menores = df[df["Age"] < 18]

# Matriz de correlación
corr = menores.corr(numeric_only=True)

# Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Matriz de correlación - Menores de edad")
plt.show()
```
